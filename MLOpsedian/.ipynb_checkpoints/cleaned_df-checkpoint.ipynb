```python
import pandas as pd

df = pd.read_csv('./data/train.csv')

# 1. Calculate the percentage of NaNs for every single column
missing_percent = (df.isnull().sum() / len(df)) * 100

# 2. Filter to show ONLY columns that actually have missing data, sorted highest to lowest
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)

print("--- Percentage of Missing Data by Column ---")
# Round to 2 decimal places for clean reading
display(missing_percent.round(2).astype(str) + ' %')

print("\n")

# 3. Calculate the overall missing percentage across the entire dataset
total_cells = df.size # Total rows * 47 columns
total_missing = df.isnull().sum().sum()
overall_missing_percent = (total_missing / total_cells) * 100

print(f"--- Overall Dataset Missingness ---")
print(f"{overall_missing_percent:.2f}% of all data points are missing.")
```

```python
# Calculate the correlation of all features against the target
correlations = df.corrwith(df['flood_risk_score'],numeric_only=True).sort_values(ascending=False)

print("--- Top 5 STRONGEST Positive Correlations ---")
display(correlations.head(5))

print("\n--- Top 5 STRONGEST Negative Correlations ---")
display(correlations.tail(5))
```
